Write a function count_vowels(text) that counts and returns the number of vowels (a, e, i, o, u) in a given sentence.

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count
```

Write a function longest_word(sentence) that returns the longest word in a given sentence

```python
def longest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest

# ask user for a sentence
sentence = input("enter a line :")
print(longest_word(sentence))
```

```python
def reverse_string(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text
    return reversed_text
#ask user for text
text = input("enter a text : ")
print(reverse_string(text))
```

Write a function is_palindrome(word) that checks if a word is a palindrome.

```python
def is_palindrome(word):
    left = 0
    right = len(word) - 1

    while left < right:
        if word[left] != word[right]:
            return False
        left += 1
        right -= 1

    return True


  #ask user for a word
word = input("enter a word : ")
print(is_palindrome(word))
if is_palindrome(word):
  print(f"palindrime")
else:
  print(f"not a palindrome")
```

Sort this dictionary:

scores = {"A": 50, "B": 10, "C": 40, "D": 30}

Without using sorted() directly.

```python
scores = {"A": 50, "B": 10, "C": 40, "D": 30}

keys = list(scores.keys())

# Manual sorting (Bubble Sort)
for i in range(len(keys)):
    for j in range(i + 1, len(keys)):
        if scores[keys[i]] > scores[keys[j]]:
            keys[i], keys[j] = keys[j], keys[i]

# Create sorted dictionary
sorted_scores = {}
for key in keys:
    sorted_scores[key] = scores[key]

print(sorted_scores)
```

Case Study: Smart Farmer – Plant Leaf Disease Advisor

A group of farmers is using a mobile app called SmartLeaf.
The app allows them to click a photo of a plant leaf, and then it automatically analyses the image and converts it into a simple Python dictionary with the following keys:

1. "spots" → color of spots on the leaf (e.g., "none", "brown", "black")

2. "patches" → color of patches on the leaf (e.g., "none", "yellow", "white")

3. "holes" → number of holes eaten by insects (integer)

4. "moisture" → "low" or "normal"

Example of one leaf report generated from the image:

leaf_report = {
    "spots": "black",
    "patches": "yellow",
    "holes": 3,
    "moisture": "low"
}


The agricultural experts have given the following simple rules to detect plant problems:

1. If there are black spots and holes ≥ 2 → disease = "Leaf Blight"

2. Else if there are yellow patches and moisture is low → disease = "Nutrient Deficiency"

3. Else if holes ≥ 4 and there are no spots → disease = "Insect Attack"

4. Otherwise → disease = "Healthy"

Also, they want to assign severity level as:

1. "High" → if disease is "Leaf Blight"

2. "Medium" → if disease is "Nutrient Deficiency" or "Insect Attack"

3. "Low" → if "Healthy"

For each disease, they also want to suggest a recommended action:

- "Leaf Blight" → "Spray fungicide and remove infected leaves"

- "Nutrient Deficiency" → "Add nitrogen-rich fertilizer and water regularly"

- "Insect Attack" → "Use organic insecticide (e.g., neem oil)"

- "Healthy" → "No action needed"

Q1) Write a Python function diagnose_leaf(leaf_report) that:

- takes a single leaf report (a dictionary) as input

- applies the above rules

- returns a new dictionary in the following format:

{
    "Disease": <disease_name_as_string>,
    "Severity": <severity_as_string>,
    "RecommendedAction": <action_as_string>
}


```python
def diagnose_leaf(leaf_report):
    # Extract values
    spots = leaf_report["spots"]
    patches = leaf_report["patches"]
    holes = leaf_report["holes"]
    moisture = leaf_report["moisture"]

    # Disease detection rules
    if spots == "black" and holes >= 2:
        disease = "Leaf Blight"
    elif patches == "yellow" and moisture == "low":
        disease = "Nutrient Deficiency"
    elif holes >= 4 and spots == "none":
        disease = "Insect Attack"
    else:
        disease = "Healthy"

    # Severity assignment
    if disease == "Leaf Blight":
        severity = "High"
    elif disease in ["Nutrient Deficiency", "Insect Attack"]:
        severity = "Medium"
    else:
        severity = "Low"

    # Recommended actions
    actions = {
        "Leaf Blight": "Spray fungicide and remove infected leaves",
        "Nutrient Deficiency": "Add nitrogen-rich fertilizer and water regularly",
        "Insect Attack": "Use organic insecticide (e.g., neem oil)",
        "Healthy": "No action needed"
    }

    return {
    "Disease": disease,
    "Severity": severity,
    "RecommendedAction": actions[disease]
    }
```

Q2) The app processes many leaf images from one field and stores them in a list of dictionaries, for example:

field_leaves = [
    {"spots": "black", "patches": "yellow", "holes": 3, "moisture": "low"},
    {"spots": "none", "patches": "yellow", "holes": 1, "moisture": "low"},
    {"spots": "none", "patches": "none", "holes": 5, "moisture": "normal"},
    {"spots": "brown", "patches": "none", "holes": 0, "moisture": "normal"}
]

Write Python code that:

- Uses the diagnose_leaf() function for each leaf

- Counts how many leaves fall under each disease category using a dictionary, for example:

{
    "Leaf Blight": 1,
    "Nutrient Deficiency": 1,
    "Insect Attack": 1,
    "Healthy": 1
}


Prints a clear summary report like:

Leaf Blight: 1 leaves

Nutrient Deficiency: 1 leaves

Insect Attack: 1 leaves

Healthy: 1 leaves

```python
field_leaves = [
    {"spots": "black", "patches": "yellow", "holes": 3, "moisture": "low"},
    {"spots": "none", "patches": "yellow", "holes": 1, "moisture": "low"},
    {"spots": "none", "patches": "none", "holes": 5, "moisture": "normal"},
    {"spots": "brown", "patches": "none", "holes": 0, "moisture": "normal"}
]

summary = {}

for leaf in field_leaves:
    result = diagnose_leaf(leaf)
    disease = result["Disease"]
    summary[disease] = summary.get(disease, 0) + 1

# Print summary report
for disease, count in summary.items():
    print(f"{disease}: {count} leaves")
```

Q3) Based on the summary, if the number of non-healthy leaves (i.e., total leaves minus healthy leaves) is more than half the total leaves, print:

"ALERT: Field at HIGH RISK. Call agricultural expert."

Otherwise, print:

"Field condition acceptable. Monitor regularly."

```python
# Sample random field data (can be changed anytime)
field_leaves = [
    {"spots": "black", "patches": "yellow", "holes": 3, "moisture": "low"},
    {"spots": "none", "patches": "none", "holes": 0, "moisture": "normal"},
    {"spots": "none", "patches": "yellow", "holes": 2, "moisture": "low"},
    {"spots": "none", "patches": "none", "holes": 5, "moisture": "normal"},
    {"spots": "brown", "patches": "none", "holes": 1, "moisture": "normal"}
]

total_leaves = len(field_leaves)
healthy_count = 0
non_healthy_count = 0

for leaf in field_leaves:
    result = diagnose_leaf(leaf)

    if result["Disease"] == "Healthy":
        healthy_count += 1
    else:
        non_healthy_count += 1

print("Total leaves:", total_leaves)
print("Healthy leaves:", healthy_count)
print("Non-healthy leaves:", non_healthy_count)

# Manual condition checking
if non_healthy_count > total_leaves / 2:
    print("ALERT: Field at HIGH RISK. Call agricultural expert.")
else:
    print("Field condition acceptable. Monitor regularly.")
```

Q4) To display a warning symbol on the app’s screen, the developers want to print the following leaf warning pattern:

Screenshot 2025-12-03 143514.png

This looks like a small leaf shape.

Write Python code using loops only to print this exact pattern.

```python
n = int(input("Enter a number: "))

# Upper half
for i in range(1, (n + 1) // 2 + 1):
    # spaces
    for j in range(1, n - i + 1):
        print(" ", end="")
    # stars (&)
    for k in range(1, (2 * i) ):
        print("*", end="")
    print()

# Lower half
for i in range((n + 1) // 2, 0, -1):
    # spaces
    for j in range(1, n - i + 1):
        print(" ", end="")
    # stars
    for k in range(1, (2 * i)):
        print("*", end="")
    print()
```
